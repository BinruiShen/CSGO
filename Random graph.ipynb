# Random graph 

```python
from scipy.spatial import Delaunay
import numpy as np

def get_adjacency_matrix(tri):
    
    edges = np.concatenate(
        (tri.vertices[:, :2], tri.vertices[:, 1:], tri.vertices[:, ::2]), axis=0
    )
    n = tri.points.shape[0]
    adj = np.zeros((n, n), dtype=np.int64)
    adj[edges[:, 0], edges[:, 1]] = 1
    return np.clip(adj + adj.T, 0, 1)


def creact_random(nPoints, conectionRate=0.5, delaunay=0, delectionRate=0, noisy=0):

    nPoints2 = int(nPoints * (1 - delectionRate))

    X = np.random.rand(nPoints, 2)  
    X2 = X[:nPoints2]

    aj = np.zeros((nPoints, nPoints))

    for i in range(nPoints):
        for j in range(i, nPoints):
            if np.random.random() < conectionRate:
                aj[i, j] = np.linalg.norm(X[i] - X[j])
                aj[j, i] = aj[i, j]
    aj2 = aj[:nPoints2, :nPoints2]

    if delaunay:
        tri1 = Delaunay(X)
        adj1 = get_adjacency_matrix(tri1)
        aj = np.multiply(adj1, aj)
        aj2 = aj[:nPoints2, :nPoints2]

    if noisy:
        for I in range(nPoints):
            index_random = np.random.randint(0, nPoints2, size=2)
            aj2[index_random[0]][index_random[1]] += np.random.random()

    return adj, adj2
```

```python
adj, adj2 = creact_random(nPoints=500, delectionRate=0.2, noisy=1, conectionRate=0.5)
```
